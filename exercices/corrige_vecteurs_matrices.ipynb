### Exercice 1 — Combinaison linéaire  <small>★☆☆☆</small>

Soient $u = (1, -2, 3)$ et $v = (4, 0, -1)$ dans $\mathbb{R}^3$. Calculer le vecteur $w = 2u - 3v$.

*Réponse attendue :* un vecteur, par exemple `Vector([a, b, c])`.

**Solution.** $w = 2(1, -2, 3) - 3(4, 0, -1) = (2, -4, 6) - (12, 0, -3) = (-10, -4, 9)$.

```python
u, v = Vector([1, -2, 3]), Vector([4, 0, -1])
Vector.combination([u, v], [2, -3])
```

Avec $u = (1, -2, 3)$ et $v = (4, 0, -1)$, calculer le produit scalaire $u^T v$ et le carré de la norme $\|u\|^2$.

*Réponse attendue :* un tuple `(produit_scalaire, norme_au_carre)`.

### Exercice 3 — Norme et vecteur unitaire  <small>★☆☆☆</small>

Soit $w = (2, -1, 2)$. Calculer $\|w\|$, puis le vecteur unitaire $\dfrac{w}{\|w\|}$ de même direction et même sens.

*Réponse attendue :* un tuple `(norme, Vector([...]))`.

**Solution.** $\|w\| = \sqrt{4 + 1 + 4} = 3$, donc $\dfrac{w}{\|w\|} = \left(\tfrac{2}{3}, -\tfrac{1}{3}, \tfrac{2}{3}\right)$.

```python
w = Vector([2, -1, 2])
w.norm(), w / w.norm()
```

### Exercice 4 — Angle entre deux vecteurs  <small>★☆☆☆</small>

Soient $a = (1, 1, 0)$ et $b = (1, 0, 1)$. Calculer $\cos\theta$, où $\theta$ est l'angle entre $a$ et $b$, puis $\theta$ (en radians, valeur exacte).

*Réponse attendue :* un tuple `(cos_theta, theta)` ; utilisez `sp.pi` pour π.

**Solution.** $a^T b = 1$, $\|a\| = \|b\| = \sqrt 2$, donc $\cos\theta = \tfrac{1}{2}$ et $\theta = \tfrac{\pi}{3}$.

```python
a, b = Vector([1, 1, 0]), Vector([1, 0, 1])
a.dot(b) / (a.norm() * b.norm()), a.angle(b)
```

```python
verifier(4, EXERCICES[4].reference)
```

### Exercice 5 — Moyenne, moyenne quadratique et écart type  <small>★☆☆☆</small>

Soit $x = (2, 4, 4, 4, 5, 5, 7, 9) \in \mathbb{R}^8$. Calculer la moyenne $\mathrm{avg}(x)$, la moyenne quadratique $\mathrm{rms}(x) = \|x\| / \sqrt{n}$ et l'écart type $\mathrm{std}(x) = \mathrm{rms}(x - \mathrm{avg}(x)\,\mathbf{1})$.

*Réponse attendue :* un tuple `(avg, rms, std)` (valeurs exactes, `sp.sqrt(...)` si besoin).

**Solution.** $\mathrm{avg}(x) = 40/8 = 5$, $\mathrm{rms}(x) = \sqrt{232/8} = \sqrt{29}$, $\mathrm{std}(x) = \sqrt{32/8} = 2$.

### Exercice 6 — Produit matrice-vecteur  <small>★★☆☆</small>

Calculer $A x$ pour $A = \begin{bmatrix} 1 & 2 & 0 \\ -1 & 3 & 2 \end{bmatrix}$ et $x = (2, -1, 1)$.

*Réponse attendue :* un vecteur.

**Solution.** Ligne 1 : $2 - 2 + 0 = 0$ ; ligne 2 : $-2 - 3 + 2 = -3$. Donc $A x = (0, -3)$.

```python
A, x = Matrix([[1, 2, 0], [-1, 3, 2]]), Vector([2, -1, 1])
A.matmul(x, view='columns')
```

### Exercice 7 — Produit matriciel et non-commutativité  <small>★★☆☆</small>

Avec $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ et $B = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$, calculer $AB$. (Calculez aussi $BA$ au brouillon : est-ce la même matrice ?)

*Réponse attendue :* la matrice $AB$.

**Solution.** $AB = \begin{bmatrix} 2 & 1 \\ 4 & 3 \end{bmatrix}$ (colonnes de $A$ échangées) alors que $BA = \begin{bmatrix} 3 & 4 \\ 1 & 2 \end{bmatrix}$ (lignes échangées) : $AB \neq BA$.

```python
A, B = Matrix([[1, 2], [3, 4]]), Matrix([[0, 1], [1, 0]])
A @ B, B @ A
```

```python
verifier(7, EXERCICES[7].reference)
```

### Exercice 8 — Transposée d'un produit  <small>★★☆☆</small>

Soient $A = \begin{bmatrix} 1 & 0 & 2 \\ 0 & 1 & -1 \end{bmatrix}$ ($2 \times 3$) et $B = \begin{bmatrix} 1 & 1 \\ 2 & 0 \\ 0 & 3 \end{bmatrix}$ ($3 \times 2$). Calculer $(AB)^T$.

*Réponse attendue :* une matrice.

**Solution.** $AB = \begin{bmatrix} 1 & 7 \\ 2 & -3 \end{bmatrix}$ donc $(AB)^T = \begin{bmatrix} 1 & 2 \\ 7 & -3 \end{bmatrix} = B^T A^T$.

```python
A = Matrix([[1, 0, 2], [0, 1, -1]])
B = Matrix([[1, 1], [2, 0], [0, 3]])
(A @ B).T, B.T @ A.T
```

```python
verifier(8, EXERCICES[8].reference)
```

### Exercice 9 — Matrice de Gram  <small>★★☆☆</small>

Pour $A = \begin{bmatrix} 1 & 1 \\ 1 & -1 \\ 2 & 0 \end{bmatrix}$, calculer la matrice de Gram $G = A^T A$. Que dire des colonnes de $A$ ?

*Réponse attendue :* la matrice $G$.

**Solution.** $G = \begin{bmatrix} 6 & 0 \\ 0 & 2 \end{bmatrix}$ : le coefficient hors diagonale est nul, les colonnes de $A$ sont **orthogonales**.

```python
A = Matrix([[1, 1], [1, -1], [2, 0]])
A.gram()
```

```python
verifier(9, EXERCICES[9].reference)
```

### Exercice 10 — Décomposer un vecteur  <small>★★☆☆</small>

Trouver les réels $\alpha, \beta$ tels que $\alpha\,(1, 2) + \beta\,(3, -1) = (7, 0)$.

*Réponse attendue :* un tuple `(alpha, beta)`.

**Solution.** $\alpha + 3\beta = 7$ et $2\alpha - \beta = 0$, donc $\beta = 2\alpha$, $7\alpha = 7$ : $\alpha = 1$, $\beta = 2$.

```python
from linalg_pedagopy import GaussJordan
GaussJordan(Matrix.from_columns([Vector([1, 2]), Vector([3, -1])]), Vector([7, 0])).run().solution
```

```python
verifier(10, EXERCICES[10].reference)
```

### Exercice 11 — Résoudre un système 3 × 3  <small>★★★☆</small>

Résoudre le système $\begin{cases} x + y + z = 6 \\ 2x - y + z = 3 \\ x + 2y - z = 2 \end{cases}$

*Réponse attendue :* le vecteur solution `Vector([x, y, z])`.

**Solution.** Par élimination : $-3y - z = -9$ et $y - 2z = -4$, d'où $z = 3$, $y = 2$, puis $x = 1$.

```python
from linalg_pedagopy import GaussJordan
res = GaussJordan(Matrix([[1, 1, 1], [2, -1, 1], [1, 2, -1]]), Vector([6, 3, 2])).run()
res.explain(level=1)
```

### Exercice 12 — Rang et noyau  <small>★★★☆</small>

Soit $A = \begin{bmatrix} 1 & 2 & 0 & 1 \\ 2 & 4 & 1 & 3 \\ 3 & 6 & 1 & 4 \end{bmatrix}$. Déterminer le rang de $A$ et une base du noyau $\ker A = \{x \in \mathbb{R}^4 : A x = 0\}$.

*Réponse attendue :* un tuple `(rang, [v1, v2, ...])` avec une liste de vecteurs.

**Solution.** $L_3 = L_1 + L_2$ : rang 2. Les pivots sont $x_1, x_3$ ; avec $x_2 = s$, $x_4 = t$ : $x_3 = -t$, $x_1 = -2s - t$. Base : $(-2, 1, 0, 0)$, $(-1, 0, -1, 1)$.

```python
A = Matrix([[1, 2, 0, 1], [2, 4, 1, 3], [3, 6, 1, 4]])
A.rank(), A.nullspace()
```

### Exercice 15 — Matrice d'une projection orthogonale  <small>★★★☆</small>

Dans $\mathbb{R}^2$, on note $p(x)$ la projection orthogonale de $x$ sur la droite $D$ d'équation $y = x$. Trouver la matrice $P$ telle que $p(x) = P x$. Vérifier ensuite (au brouillon) que $P^2 = P$.

*Réponse attendue :* une matrice $2 \times 2$.

**Solution.** $u = (1, 1)$, $P = \dfrac{u u^T}{u^T u} = \tfrac{1}{2}\begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$ ; on vérifie $P^2 = P$ et $P^T = P$.

```python
u = Vector([1, 1])
P = Matrix.from_columns([u]) @ Matrix.from_rows([u]) / u.norm_sq()
P, P @ P == P
```

```python
verifier(15, EXERCICES[15].reference)
```

### Exercice 17 — Orthonormalisation de Gram-Schmidt  <small>★★★★</small>

Appliquer le procédé de Gram-Schmidt à $a_1 = (1, 1, 0)$, $a_2 = (1, 0, 1)$, $a_3 = (0, 1, 1)$ (dans cet ordre) pour obtenir une base orthonormée $(q_1, q_2, q_3)$.

*Réponse attendue :* un tuple `(q1, q2, q3)` de vecteurs (utilisez `sp.sqrt`).

**Solution.** $q_1 = \tfrac{1}{\sqrt 2}(1, 1, 0)$ ; $\tilde q_2 = a_2 - \tfrac12(1, 1, 0) = \tfrac12(1, -1, 2)$, $q_2 = \tfrac{1}{\sqrt 6}(1, -1, 2)$ ; $\tilde q_3 = a_3 - \tfrac12(1,1,0) - \tfrac16(1,-1,2) = \tfrac23(-1, 1, 1)$, $q_3 = \tfrac{1}{\sqrt 3}(-1, 1, 1)$.

```python
from linalg_pedagopy import GramSchmidt
res = GramSchmidt([Vector([1, 1, 0]), Vector([1, 0, 1]), Vector([0, 1, 1])]).run()
res.orthonormal
```

```python
verifier(17, EXERCICES[17].reference)
```

### Exercice 19 — Droite des moindres carrés  <small>★★★★</small>

On cherche la droite $y = c_0 + c_1 t$ qui approche au mieux, au sens des moindres carrés, les points $(0, 1)$, $(1, 3)$, $(2, 4)$, $(3, 4)$. Calculer $\hat c = (c_0, c_1)$.

*Réponse attendue :* le vecteur `Vector([c0, c1])`.

**Solution.** $A^T A = \begin{bmatrix} 4 & 6 \\ 6 & 14 \end{bmatrix}$, $A^T y = (12, 23)$, d'où $\hat c = (3/2,\ 1)$ : la droite $y = 3/2 + t$.

### Exercice 20 — Puissances d'une matrice  <small>★★★★</small>

Soit $A = \begin{bmatrix} 2 & 1 \\ 0 & 2 \end{bmatrix}$. Exprimer $A^n$ en fonction de l'entier $n \ge 0$ (utilisez le symbole `n = symbols('n')[0]`).

*Réponse attendue :* une matrice contenant le symbole `n`.

**Solution.** $A = 2I + N$ avec $N^2 = 0$, donc $A^n = 2^n I + n\,2^{n-1} N = \begin{bmatrix} 2^n & n\,2^{n-1} \\ 0 & 2^n \end{bmatrix}$.

```python
A = Matrix([[2, 1], [0, 2]])
[A @ A, A @ A @ A]   # on devine la forme, puis on la démontre
```
